```python

print("NAME: SHALINI A")
print("ROLL NO: 24BAD409")
print("EXPERIMENT NO: 9")
print("TEXT CLASSIFICATION AND TEXT GENERATION USING TRANSFORMERS")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from torch.utils.data import DataLoader
from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    AutoModelForSeq2SeqLM,
    DataCollatorWithPadding)
from torch.optim import AdamW
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("\nDevice:", device)


# =====================================================
# PART A: DATASET PREPARATION AND TEXT PREPROCESSING
# =====================================================

print("\nPART A: DATASET PREPARATION")

dataset = load_dataset("stanfordnlp/imdb")

print("\nDataset Information:")
print(dataset)

print("\nSample Review:")
print(dataset["train"][0]["text"])

print("\nSample Label:", dataset["train"][0]["label"])

print("\nTraining Samples:", len(dataset["train"]))
print("Testing Samples:", len(dataset["test"]))

print("\nClass Distribution:")
print(dataset["train"].to_pandas()["label"].value_counts())

plt.figure(figsize=(6, 4))
dataset["train"].to_pandas()["label"].value_counts().sort_index().plot(
    kind="bar"
)
plt.title("IMDb Sentiment Distribution")
plt.xlabel("Sentiment (0 = Negative, 1 = Positive)")
plt.ylabel("Number of Reviews")
plt.show()


# Select a smaller dataset for faster training
TRAIN_SAMPLES = 5000
TEST_SAMPLES = 1000

train_data = dataset["train"].shuffle(seed=42).select(
    range(TRAIN_SAMPLES))

test_data = dataset["test"].shuffle(seed=42).select(
    range(TEST_SAMPLES))

split_data = train_data.train_test_split(
    test_size=0.1,
    seed=42)
train_data = split_data["train"]
validation_data = split_data["test"]


# Load BERT tokenizer
bert_tokenizer = AutoTokenizer.from_pretrained(
    "google-bert/bert-base-uncased")

def tokenize_function(examples):
    return bert_tokenizer(
        examples["text"],
        truncation=True,
        max_length=128)

train_tokenized = train_data.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)

validation_tokenized = validation_data.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"]
)

test_tokenized = test_data.map(
    tokenize_function,
    batched=True,
    remove_columns=["text"])

data_collator = DataCollatorWithPadding(
    tokenizer=bert_tokenizer)

train_loader = DataLoader(
    train_tokenized,
    batch_size=16,
    shuffle=True,
    collate_fn=data_collator)

validation_loader = DataLoader(
    validation_tokenized,
    batch_size=16,
    collate_fn=data_collator)

test_loader = DataLoader(
    test_tokenized,
    batch_size=16,
    collate_fn=data_collator)

print("\nText preprocessing completed successfully")
print("Training batches:", len(train_loader))
print("Validation batches:", len(validation_loader))
print("Testing batches:", len(test_loader))

# =====================================================
# PART B: FINE-TUNING BERT FOR TEXT CLASSIFICATION
# =====================================================
print("\nPART B: BERT TEXT CLASSIFICATION")
bert_model = AutoModelForSequenceClassification.from_pretrained(
    "google-bert/bert-base-uncased",
    num_labels=2)
bert_model.to(device)
optimizer = AdamW(
    bert_model.parameters(),
    lr=2e-5
)
EPOCHS = 2
training_losses = []
validation_losses = []


# Evaluation function
def evaluate_model(model, loader):
    model.eval()
    total_loss = 0
    predictions = []
    actual_labels = []
    with torch.no_grad():
        for batch in loader:
            batch = {
                key: value.to(device)
                for key, value in batch.items()
            }
            outputs = model(**batch)
            total_loss += outputs.loss.item()
            preds = torch.argmax(
                outputs.logits,
                dim=1
            )

            predictions.extend(preds.cpu().numpy())
            actual_labels.extend(
                batch["labels"].cpu().numpy()
            )
    avg_loss = total_loss / len(loader)
    return avg_loss, predictions, actual_labels

# Train BERT model
print("\nStarting BERT Training...")
for epoch in range(EPOCHS):
    bert_model.train()
    total_loss = 0
    progress = tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS}"
    )
    for batch in progress:
        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }
        optimizer.zero_grad()

        outputs = bert_model(**batch)
        loss = outputs.loss
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        progress.set_postfix(loss=loss.item())
    avg_train_loss = total_loss / len(train_loader)
    val_loss, _, _ = evaluate_model(
        bert_model,
        validation_loader
    )
    training_losses.append(avg_train_loss)
    validation_losses.append(val_loss)
    print("\nEpoch:", epoch + 1)
    print("Training Loss:", avg_train_loss)
    print("Validation Loss:", val_loss)
print("\nBERT Training Completed Successfully")


# Plot training and validation loss
plt.figure(figsize=(8, 5))
plt.plot(
    training_losses,
    marker="o",
    label="Training Loss"
)
plt.plot(
    validation_losses,
    marker="o",
    label="Validation Loss"
)
plt.title("BERT Training and Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()


# Evaluate using testing dataset
test_loss, predictions, actual_labels = evaluate_model(
    bert_model,
    test_loader
)
accuracy = accuracy_score(actual_labels, predictions)
precision = precision_score(
    actual_labels,
    predictions,
    zero_division=0
)
recall = recall_score(
    actual_labels,
    predictions,
    zero_division=0
)
f1 = f1_score(
    actual_labels,
    predictions,
    zero_division=0)

print("\nBERT MODEL EVALUATION")
print("Test Loss:", test_loss)
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)

print("\nClassification Report:")
print(
    classification_report(
        actual_labels,
        predictions,
        target_names=["Negative", "Positive"],
        zero_division=0)
)


# Test BERT using new text inputs
print("\nTESTING NEW TEXT INPUTS")
sample_reviews = [
    "This movie was amazing and I enjoyed every moment.",
    "The movie was boring and completely disappointing.",
    "The acting was excellent and the story was wonderful.",
    "I did not like this movie. It was a waste of time."
]
bert_model.eval()
for review in sample_reviews:
    inputs = bert_tokenizer(
        review,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=128
    ).to(device)
    with torch.no_grad():

        outputs = bert_model(**inputs)

        predicted_class = torch.argmax(
            outputs.logits,
            dim=1
        ).item()
    sentiment = "Positive" if predicted_class == 1 else "Negative"
    print("\nReview:", review)
    print("Predicted Sentiment:", sentiment)

# Save BERT model and tokenizer
bert_model.save_pretrained("bert_movie_classifier")
bert_tokenizer.save_pretrained("bert_movie_classifier")
print("\nBERT Model and Tokenizer Saved Successfully")

# =====================================================
# PART C: SEQUENCE-TO-SEQUENCE TEXT GENERATION USING T5
# =====================================================
print("\nPART C: TEXT GENERATION USING T5")
t5_model_name = "google-t5/t5-small"
t5_tokenizer = AutoTokenizer.from_pretrained(
    t5_model_name
)
t5_model = AutoModelForSeq2SeqLM.from_pretrained(
    t5_model_name
)
t5_model.to(device)
print("\nT5 Model Loaded Successfully")

# Text generation function
def generate_text(text, max_new_tokens=60, num_beams=4):
    input_text = "summarize: " + text
    inputs = t5_tokenizer(
        input_text,
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)
    with torch.no_grad():
        outputs = t5_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            num_beams=num_beams,
            early_stopping=True
        )

    generated_text = t5_tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return generated_text


# Sample input texts
sample_texts = [
    "Artificial intelligence is a branch of computer science that develops machines capable of performing tasks that normally require human intelligence. It is used in healthcare, education, transportation and many other fields.",

    "Machine learning is a part of artificial intelligence. It allows computers to learn from data and make predictions without being explicitly programmed. It is widely used in recommendation systems, fraud detection and image recognition.",

    "Climate change is a major environmental problem. Increasing global temperatures affect rainfall, agriculture, biodiversity and human health. Renewable energy and environmental protection can help reduce its impact."
]


# Generate text for different inputs
generated_outputs = []
for i, text in enumerate(sample_texts):
    generated = generate_text(
        text,
        max_new_tokens=60,
        num_beams=4
    )
    print("\nInput Text", i + 1)
    print(text)
    print("\nGenerated Output:")
    print(generated)
    generated_outputs.append({
        "Input": text,
        "Generated Output": generated
    })

# Experiment with different generation settings
print("\nCOMPARING GENERATION SETTINGS")
comparison_text = sample_texts[0]
output_1 = generate_text(
    comparison_text,
    max_new_tokens=30,
    num_beams=1
)
output_2 = generate_text(
    comparison_text,
    max_new_tokens=60,
    num_beams=4
)
print("\nGeneration with 30 Tokens and 1 Beam:")
print(output_1)
print("\nGeneration with 60 Tokens and 4 Beams:")
print(output_2)

# Save generated outputs

output_df = pd.DataFrame(generated_outputs)
output_df.to_csv(
    "generated_text_outputs.csv",
    index=False
)
print("\nGenerated Text Outputs Saved Successfully")
# Save T5 model and tokenizer
t5_model.save_pretrained("t5_text_generation")
t5_tokenizer.save_pretrained("t5_text_generation")
print("T5 Model and Tokenizer Saved Successfully")


# =====================================================
# PART D: MODEL TESTING AND OUTPUT ANALYSIS
# =====================================================
print("\nPART D: MODEL TESTING AND OUTPUT ANALYSIS")
print("\nBERT CLASSIFICATION RESULTS")
for i in range(len(sample_reviews)):
    print("\nReview:", sample_reviews[i])
    print(
        "Predicted:",
        "Positive" if bert_model(
            **bert_tokenizer(
                sample_reviews[i],
                return_tensors="pt",
                truncation=True,
                max_length=128
            ).to(device)
        ).logits.argmax(dim=1).item() == 1 else "Negative"
    )

print("\nT5 TEXT GENERATION RESULTS")
for i, item in enumerate(generated_outputs):
    print("\nSample:", i + 1)
    print("Input:", item["Input"])
    print("Generated:", item["Generated Output"])

# Final comparison
print("\nMODEL COMPARISON")

print("BERT:")
print("Used for text classification and sentiment prediction.")

print("\nT5:")
print("Used for sequence-to-sequence text generation and summarization.")

print("\nFINAL RESULT:")
print("BERT text classification completed successfully.")
print("T5 text generation completed successfully.")
print("Models evaluated and outputs saved successfully.")
```
